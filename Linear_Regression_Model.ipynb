```python
# imports
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# create data
np.random.seed(0)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

# split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

```python
# train model
model = LinearRegression()
model.fit(X_train, y_train)

# evaluate model
y_pred = model.predict(X_test)
print("Coefficients:", model.coef_)
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred))
print("Coefficient of determination (R^2): %.2f" % r2_score(y_test, y_pred))
```

```python
# visualize
plt.scatter(X, y, color="black")
plt.plot(X_test, y_pred, color="blue", linewidth=3)
plt.title("Linear Regression Line vs Actual Data")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

```python
# explain accuracy
explanation = """
Simple Explanation of Model Accuracy:
1. The R^2 score tells us how well the independent variable explains the changes in the dependent variable.
2. The Mean Squared Error measures the average squared difference between the estimated values and the actual value.
"""
print(explanation)
```
